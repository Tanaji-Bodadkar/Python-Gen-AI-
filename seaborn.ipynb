## 3. Matrix Plot
#### relationship between 2d like table using colors to show the strength of pattern of data

### a. Heatmap

#### In Every Classification Machine Learning Problem Statement , we use the Heatmap .

#### Heatmap uses color shades to show the size or Numbers in a Matrix 2D table
#### Darker or Lighter colors indicate higher or lower values .
#### Hyper Useful when we want to see correlation , Missing Data , comparisons , density

#### The Matrix Data must be Table in Different Categories
#### The Matrix Data must be Table in Different Categories

#### for example as an Heatmap or as Matrix which we can convert into an Heatmap , we create a Pivot Table and then that Pivot Table only we can use to create an Heatmap .

#### pivot()---> index, columns, values ---> gap.pivot(index='', columns='', values='')

#### in matirx plot we are not having figure level 
#### only axis level i.e. heatmap()-data,annot,cmap

```python
gap.sample(5)
```

```python
dt=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(data=dt)
```

```python
dt = gap[gap['continent']== 'Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(data=dt,annot=True,cmap='viridis')
```
