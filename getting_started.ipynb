#### 2d line plot - usual works with 2 columns - categorical,numerical

```python
df.head()
```

```python
data = df['Gender'].value_counts()
```

```python
x = data.index
y = data.values
```

```python
plt.plot(x,y)
```

#### plt.figure() - parameter - figsize - figure dimension of our graph
#### plt.figure(figsize=(10,6))

#### plt.title() - title() - title of our plot - parameter,(color of title)

#### plt.ylabel() - represents what y axes is representing 
#### plt.xlabel() - represents what x axes is representing

#### plt.grid() - brings grid on the plot

#### plt.show() - represents the chart

#### plt.xlim() -> [100,1000] - x values
#### plt.ylim() -> [same as x] - y values

```python
plt.figure(figsize=(10,4))
plt.title('Count of population as per gender')
plt.ylabel('Count of pop')
plt.xlabel('Gender')
plt.plot(x,y,color='red',marker='o',linestyle='-.',linewidth=3,markersize=7,label='Gender population')
plt.grid()
plt.show()
```

```python
sales = pd.read_csv('Chocolate Sales.csv')
```

```python
sales
```

```python
sales.head()
```

```python
sales.info()
```

```python
sales['Date']=pd.to_datetime(sales['Date'])
```

```python
sales['Date']=sales['Date'].dt.month_name()
```

```python
sales['Date']
```

```python
sales['Sales Person']=sales['Sales Person'].astype('category')
sales['Country']=sales['Country'].astype('category')
sales['Product']=sales['Product'].astype('category')
sales['Date']=sales['Date'].astype('category')
```

```python
sales['Amount']=sales['Amount'].str.replace(",","")
```

```python
sales['Amount'] = sales['Amount'].str.replace(",", "").str.replace("$", "").str.strip().astype('int32')
```

```python
sales.info()
```

```python
sales['Boxes Shipped']=sales['Boxes Shipped'].astype('int32')
```

```python
sales.info()
```

```python
data=sales.groupby('Date')['Amount'].sum()
```

```python
data.index=pd.CategoricalIndex(data.index,categories=['January','February','March','April','May','June','July','August'])
```

```python
data=data.sort_index()
```

```python
x = data.index
y = data.values
```

```python
plt.figure(figsize=(10,5))
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Amount')
plt.grid()
plt.plot(x,y,color='green',linestyle='--',marker='o')
plt.show()
```
