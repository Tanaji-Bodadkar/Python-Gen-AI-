## 2D Line Plot
#### Specific purpose -> specific purpose
#### 2d line plot - usual works with 2 columns - categorical,numerical 
#### x-axes -> category column
#### y-axes -> numerical column
#### It can be used for Time Series Data

```python
df.head()
```

```python
data = df['Gender'].value_counts()
```

```python
x = data.index
y = data.values
```

```python
plt.plot(x,y)
```

#### plt.figure() - parameter - figsize - figure dimension of our graph
#### plt.figure(figsize=(10,6))

#### plt.title() - title() - title of our plot - parameter,(color of title)

#### plt.ylabel() - represents what y axes is representing 
#### plt.xlabel() - represents what x axes is representing

#### plt.grid() - brings grid on the plot

#### plt.show() - represents the chart

#### plt.xlim() -> [100,1000] - x values
#### plt.ylim() -> [same as x] - y values

```python
plt.figure(figsize=(10,4))
plt.title('Count of population as per gender')
plt.ylabel('Count of pop')
plt.xlabel('Gender')
plt.plot(x,y,color='red',marker='o',linestyle='-.',linewidth=3,markersize=7,label='Gender population')
plt.grid()
plt.show()
```

#### customization of line
#### parameters : color , marker , linestyle , linewidth , markersize , label
