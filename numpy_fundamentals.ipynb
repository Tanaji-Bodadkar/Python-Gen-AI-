## NUMPY
#### NumPy is a Fundamental Package for scientific computing in Python. It is a Python Library that provides a Multi-Dimensional array object and various derived objects such as Masked Arrays or Matrices and an assortment of routines for fast operations on Arrays including Mathematical, Logical, Shape Manipulation, Sorting, Selecting, I/O, Discrete Fourier Transforms, Basic Linear Algebra, Statistical Operations, Random Simulation, etc.
#### this package is the ndimensional Array object

#### Numpy Arrays vs Python Sequences 

#### Numpy Arrays fixed size, unlike python lists which grow dynamically. changing the size of ndim array will create new array and delete original

#### The elements in numpy array are required to be of the same data type same size and meomry 

#### numoy arrays facilitate advanced mathematical and othe type of operation on large number of data , typically such operations are executed efficiently with less code than is possible using pythons built in sequences.
