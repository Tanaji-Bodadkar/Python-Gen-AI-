```python
price_dict["Lindt"] = 196
print(price_dict)
```

#### DELETE A product


```python
product = "Lays"
if product in price_dict:
    del price_dict[product]
    print(product, "removed successfully ")
else:
    print("product does not exists")

```

### 3 Average


```python
total = sum(price_dict.values())
count =  len(price_dict)
average = total / count 
print(average)
```

### Task 4 Combined Opertaions


#### 1 Using the products list and price_dict, create a list of tuples named catalog where each tuple is (product_name, price, category).

```python
products = ["Cadbury", "Lays", "Mixtures", "Maggie", "Lindt", "Nestle"]

categories = [
    "Chocolate",
    "Chips",
    "Snacks",
    "Noodles",
    "Chocolate",
    "Chocolate"
]

price_dict = {
    "Cadbury": 55,
    "Lays": 20,
    "Mixtures": 80,
    "Maggie": 15,
    "Lindt": 250,
    "Nestle": 60
}
```

```python
catalog = []
for i in range(len(products)):
    product = products[i]
    price = price_dict[product]
    category = categories[i]

    catalog.append((product,price,category))

print(catalog) 
```

#### 2 From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.

```python
category_to_products = {}

for product, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []

    category_to_products[category].append(product)

print(category_to_products)
```

#### 3. Find the category with the maximum number of products

```python
max_category = max(category_to_products, key=lambda category: len(category_to_products[category]))

print("Category with maximum products:", max_category)
print("Products:")

for product in category_to_products[max_category]:
    print(product)
```
