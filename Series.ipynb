#### Series is a 1d array that can hold any type of data like int,float,strings(obj),boolean,etc.

#### Data type ---> Integers ----> Data in a Series(Column) ---> Quantity of items in the Stock
#### Data type ---> Float ----> Data in a Series(Column) ---> Price or Product Ratings
#### Data type ---> Strings ---> Column ---> Product Name
#### Data type ---> Boolean ---> Column ---> Is item in Stock ?
#### Data type ---> Dates/ Times ---> Column ---> Expiry Dates or Sale Dates
#### Mixed Types ---> Milk, 40, True

#### Creating a series in multiple ways
#### 1 List
#### 2 Numpy array
#### 3 Dictionary
#### 4 Custom index
#### 5 Scalar value

```python
## Series from list
country = ['INDIA','USA','JAPAN','CHINA']
pd.Series(country)
```

```python
batsman_score=[34,4,56,29,58]
pd.Series(batsman_score,dtype='int32')
```

```python
subjects=['MATHS','PHYSISCS','CHEMISTRY','ENGLISH']
scores=[79,80,45,90]
pd.Series(data=scores,index=subjects,name='Subject Marks Data',dtype='int32')
```

```python
## Creating a series from numpy array
values = np.arange(0,11)
pd.Series(values)
```

```python
## dictionary to series
## dictionart is a another data structure that stores  the data in a key value pair so we dont have to explicitly mention the index
```

```python
data_items={'Physics':78,'Chemistry':88,'Maths':92}
pd.Series(data_items)
```

```python
## custom index
values = [1,2,3,4,5]
pd.Series(data=values,index=['a','b','c','d','e'])
```

```python
## scalae values inside a series 
pd.Series(4,index=['a','b','c','d','e'])
```

### Mathematical Functions in Pandas
#### 1. sum() -> Sum of all values present in the series
#### 2. product() -> product of all values in the series
#### 3. mean() -> it provides avg value of data
#### 4. median() -> middle vaalue of data afte sorting it to acsending order
#### 5. mode() -> maximum repeated value
#### 6. std() -> distribution of values in series form mean values
#### 7. var() -> square of std
#### 8. min() & max()
#### 9. describe() -> provides the summary about the imp mathematical calculations for the data it will give minimum value and maximum value, mean , std, 25%ile quartile range , count-total rows in column
