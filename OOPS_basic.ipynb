## OOPS- Classes and objects consisting of blueprints

```python
def __init__(self,name,power,weapon,nationality):
    self.name=name
    self.power=power
    self.weapon = weapon
    self.nationality =nationality

def introduction(self):
    print(f"Hello my name is {self.name}")

```

```python
class Avenger:
    def __init__(self,name,power,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"hello my name is {self.name}")

    def carrying_weapon(self):
        print(f"I am carryin the weapon {self.weapon}")

```

```python
Ironman = Avenger("Iron man","Genius","Repulsor beam")
print(Ironman.introduction())
print(Ironman.carrying_weapon())
```
