#### Encapsulation 
##### it is a way of wrapping data inside a capsule or a class as single unit 
##### key concepts:
##### Public members: accessible from anywhere  , 
##### Protected members: data is not accessible outside the class but accessible by derived class or child class, represented by using single underscore
 
##### Private members: is not acccessible outside the class and not even in the derived class in order to access them or modify we use getter and setter method* it is declared using double underscore__ before the name  
##### & getter and setter method

```python
class Ironman:
    def __init__(self,name,power):
        self.name= name
        self.power=power

    def intro(self):
        print(f"Hello my name is {self.name} and i have {self.power} power.")
                
```

```python
iron=Ironman("IRONMAN","BEAM")
dir(iron)
```

```python
iron.intro()
```

```python
## getter method -> it will be helpin us out with accessing the private data

class Captain:
    def __init__(self,name,weapon):
        self.name=name
        self.__weapon=weapon

    ##getter method
    def accessing_weapon(self):
        print(f"my weapon is {self.__weapon}")
```

```python
captain = Captain("Captain America","Shielf")
```

```python
captain.accessing_weapon()
```

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.__sound = sound

    def sound_info(self):
        print(f"the animal is going to sound as {self.__sound}")

    def modify_sound(self, new_sound):
        self.__sound = new_sound  # Update the private attribute
        print(f"new sound for animal is: {self.__sound}")  # Access the private attribute
```

```python
animal = Animal("Bruno","Bark")
```

```python
animal.sound_info()
```

```python
animal.modify_sound("meow")
```

```python
## protected data 
class Grandparent:
    def __init__(self, name, locker_key):
        self.name = name
        self.__locker_key = locker_key

    def get_locker_key(self):
        return self.__locker_key


class Parent(Grandparent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)

    def child_reveal(self):
        print(f"The locker key is: {self.get_locker_key()}")


parent = Parent("Amit", "123456")
parent.child_reveal()
```
