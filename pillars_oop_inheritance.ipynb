## Pillars of OOP

### Inheritance 
### It is a fundamental concept in OOP because of which child class can inherit the properties or attributes and functionalities of parent class 

### In python - 3 types of Inheritance 
### 1. Single 
### 2. Multilevel
### 3. Multiple  

#### Single Inheritance 

```python
class Avenger:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    def intro(self):
        print(f"Hello my name is {self.name}")

class Ironman(Avenger):
    def __init__(self, name, power,weapon):
        super().__init__(name, power)
        self.weapon=weapon

    def weapono(self):
        print(f"i use {self.weapon}")


```

```python
iron =  Ironman("Ironman","Genius Intellect","Blue beam")
```

#### Multilevel : Grandparent -> Parent -> Child 

```python
class Grandparent:
    def __init__(self,nationality,baldness,anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def introduction(self):
        print(f"Hello my nationality is {self.nationality} and about my anger_issues are {self.anger_issues} , i am bald {self.baldness}")
        


        
```

```python
class Parent(Grandparent):

    def __init__(self, nationality, baldness , anger_issues, intelligence):
        super().__init__(nationality,baldness,anger_issues)
        self.intteligence=intelligence

    def intelligence_details(self):
        print(f"i am intelligent yes or no {self.intelligence}")
```

```python
class Child(Parent):
    def __init__(self,  nationality, baldness , anger_issues, intelligence,charm):
        super().__init__(nationality, baldness , anger_issues, intelligence)
        self.charm=charm

    def cham_in(self):
        print(f"i am charmin yes or no {self.charm}")
```

```python
parent_1 = Parent("Indian","Yes","No","Yes")
```
