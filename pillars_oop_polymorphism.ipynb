### Polymorphism
##### means many forms in different class in which object is treated same all over


#### Types :
#### 1. Method Overriding  (Runtime polymorphism)
#### -> a child class provides a specific implementation of a method that has already been defined into the parent class
#### 2. Method Overloading (Compiletime polymrphism)
#### -> 
#### 3. Operator Overloading 

```python
class Grandparent:
    def speak(self):
        print("Hello my name is Tanaji")

class Parent(Grandparent):
    def speak(self):
        print("Hello my name is Tanaji Jr.")

class Child(Grandparent):
    def speak(self):
        print("hello i am Tanaji's Grandchild")

```

```python
def call(objects):
    return objects.speak()
```

```python
Grandpa=Grandparent()
papa = Parent()
```

```python
call(Grandpa)
```

```python
call(papa)
```

```python
class Shape:
    def area(self):
        return "This method returns the area of shape"
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self):
        return self.length*self.width

class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius*self.radius
```

```python
shape = Shape()
shape.area()
```

```python
rectangle = Rectangle(6,4)
rectangle.area()
```

```python
circle = Circle(5)
circle.area()
```
