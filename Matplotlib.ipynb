#### plt.xlabel() - represents what x axes is representing

#### Multiple Bar Charts

```python
sales.head()
```

```python
m = sales[sales['Product']=='Mint Chip Choco']	
d = sales[sales['Product']=='85% Dark Bars']
p = sales[sales['Product']=='Peanut Butter Cubes']	
		
```

```python
sales['Date'] = pd.to_datetime(sales['Date'], format='%d-%b-%y')
```

```python


m = sales[sales['Product']=='Mint Chip Choco']	
d = sales[sales['Product']=='85% Dark Bars']
p = sales[sales['Product']=='Peanut Butter Cubes']	

m = m.groupby(m['Date'].dt.month_name())['Amount'].sum()
d = d.groupby(d['Date'].dt.month_name())['Amount'].sum()
p = p.groupby(p['Date'].dt.month_name())['Amount'].sum()
```

```python
m
```

```python
months = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

m.index = pd.CategoricalIndex(
    m.index,
    categories=months,
    ordered=True
)

m = m.sort_index()
```

```python
months = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

d.index = pd.CategoricalIndex(
    d.index,
    categories=months,
    ordered=True
)

d = d.sort_index()
```

```python
months = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

p.index = pd.CategoricalIndex(
    p.index,
    categories=months,
    ordered=True
)

p = p.sort_index()
```

```python
d
```

```python
plt.figure(figsize=(9,5))

plt.title('Month / Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')

x = np.arange(m.shape[0])

plt.bar(x - 0.2, m.values,
         color='green',
         label='Mint Chip Choco',
         width=0.2)

plt.bar(x, d.values,
         color='blue',
         label='85% Dark Bars',
         width=0.2)

plt.bar(x + 0.2, p.values,
         color='red',
         label='Peanut Butter Cubes',
         width=0.2)

plt.xticks(x, m.index,rotation='vertical')

plt.legend()
plt.show()
```

### Stacked Bar Chart

```python
plt.figure(figsize=(9,5))

plt.title('Month / Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')

x = np.arange(m.shape[0])

plt.bar(m.index, m.values,
         color='green',
         label='Mint Chip Choco')

plt.bar(d.index, d.values,
         color='blue',
         label='85% Dark Bars',bottom=m.values)

plt.bar(p.index, p.values,
         color='red',
         label='Peanut Butter Cubes',bottom=m.values+d.values)

plt.legend()
plt.show()
```
