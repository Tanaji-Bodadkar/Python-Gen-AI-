## Magic Methods

## the method can help us define
#### 1.Object initialization(__init__)
#### 2.String  representation(__str__,__repr__)
#### 3.Arithmetic operations(__add__,__sub__,etc)
#### 4.Comparison Operations(eq,lt)
#### 5. Length , Indexing iteration(len,getitem,etc)

##  why use:
#### 1 make object behvae like builtin types
#### 2 improves readability 
#### 3 Simplifies code by replacing verbose function calls with natural syntax

```python
##__add__
class Avenger:
    def __init__(self,name,strength):
        self.name=name
        self.strength=strength

    def __add__(self,other):
        return Avenger(self.name + " & " + other.name,self.strength + other.strength)

    def __str__(self):
        return f"{self.name} with combine strength {self.strength}"

ironman=Avenger("Ironman",80)
thor = Avenger("Thor",120)

team = ironman + thor
print(team)
```
