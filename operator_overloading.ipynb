## Operator Overloading 

### it allows us to to define how operators +,-,/,*,%,etc work with user defined objects.
#### why is operator overloading useful?
#### - makes custom objects behave like built in types
#### - improves code readability by using natural syntax istead object.add


#### Operator '+' -> Python calls __add__(self, other)
#### Operator '-' -> Python calls __sub__(self, other)
#### Operator '*' -> Python calls __mul__(self, other)
#### Operator '/' -> Python calls __truediv__(self, other)
#### Operator '==' -> Python calls __eq__(self, other)
#### Operator '>' -> Python calls __gt__(self, other)
#### Operator '//' -> Python calls __floordiv__(self, other)
#### Operator '%' -> Python calls __mod__(self, other)

```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power = power

    def __add__(self, other):
        return Avenger(self.name + " & " + other.name,
                        self.power + other.power)

    def __sub__(self, other):
        return Avenger(self.name,
                        self.power - other.power)

    def __mul__(self, other):
        return Avenger(self.name,
                        self.power * other.power)

    def __str__(self):
        return f"{self.name} with power {self.power}"
```

```python
thor=Avenger("Thor",100)
```

```python
hulk=Avenger("Hulk",120)
```

```python
team= thor + hulk
```
