#### Abstraction : it is a concept of hiding the implementation details and only showing the necessary features of the object.
#### this reduces programming complexity and efforts 
#### how can we acheive abstraction in python
#### 1.Abstract class -> a class that cannot be instantiated or that means for an abstract class we cant create the object of this class
#### 2.Abstract method:A method declared inside an abstract class that has no implementation (body). Any subclass inheriting from the abstract class must override and implement this method.
#### define by @abstractmethod

```python
from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("The car starts with an ignition key")
    def stop(self):
        print("It will stop using break padals")

class ElectricScooter:

    def start(self):
        print("Electric scooter will start by power button")

    def stop(self):
        print("Electric scooter stops with an electric break")

```

```python
car = Car()
```

```python
car.start()
```

```python
scooter = ElectricScooter()
```

```python
scooter.start()
```
